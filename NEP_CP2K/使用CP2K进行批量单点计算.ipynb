# 使用 CP2K 进行批量单点计算
## 使用 ASE 拆分训练集

````python
from ase.io import iread, write
import os

output_folder = "./GeTe_2024"
os.makedirs(output_folder, exist_ok=True)
nlist = []
i = 0
for atoms in iread("./train.xyz"):
    natoms = len(atoms)
    if natoms not in nlist:
        nlist.append(natoms)
        sub_folder = f"{output_folder}/{natoms}"
        os.makedirs(sub_folder, exist_ok=True)       
    if "forces" in atoms.arrays:
        del atoms.arrays["forces"]
    write(f"./{output_folder}/{natoms}/{i}.cif", atoms, format="cif")
    i += 1
````

```shell
cd $1
```
