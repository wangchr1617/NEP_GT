# CP2K 构建非晶模型

## 使用 Packmol 生成初始结构

`generate.inp` 文件内容如下：

```
tolerance 2 # 容忍的原子间距下限
output ini.pdb # 指定输出文件
filetype xyz # 输入的结构文件的文件格式
seed -1 # 随机数种子，-1 表示不固定
structure model.xyz # 以 model.xyz 作为输入的结构文件
  number 150 # 重复单元数
  inside cube 0. 0. 0. 16.5759 # 晶胞是一个边长为 16.5759 埃的盒子
end structure
```

使用命令 `packmol < generate.inp` 生成初始结构。
