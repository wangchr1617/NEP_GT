```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import scienceplots
from scipy.interpolate import CubicSpline
from scipy.optimize import curve_fit
from scipy.optimize import minimize

plt.style.use(['science', 'ieee', 'no-latex', 'bright'])
colors = sns.color_palette("rocket")

def extract_energy_per_atom_from_xyz(file_path):
    energy_per_atom_list = []
    with open(file_path, 'r') as file:
        lines = file.readlines()
    i = 0
    while i < len(lines):
        num_atoms = int(lines[i].strip())
        i += 1
        if "energy=" in lines[i]:
            energy_line = lines[i]
            energy_value = float(energy_line.split("energy=")[1].split()[0])
            energy_per_atom = energy_value / num_atoms
            energy_per_atom_list.append(round(energy_per_atom, 8))
        i += num_atoms + 1
    return energy_per_atom_list
    
def zbl_function(x, params):
    a1, a2, a3, a4, a5, a6, a7, a8 = params
    return (a1 * np.exp(-a2 * x) + a3 * np.exp(-a4 * x) + a5 * np.exp(-a6 * x) + a7 * np.exp(-a8 * x))

def objective_function(params, x_data, y_data):
    y_pred = zbl_function(x_data, params)
    error = y_data - y_pred
    return np.mean(error ** 2)
```

```python
gege = extract_energy_per_atom_from_xyz("./ivdw0/Ge_Ge.xyz")
gete = extract_energy_per_atom_from_xyz("./ivdw0/Ge_Te.xyz")
tete = extract_energy_per_atom_from_xyz("./ivdw0/Te_Te.xyz")

len(gege), len(gete), len(tete)
```

```python
initial_guess = np.array([[544.58631578, 132.4849635, 1697.63672345, 195.41641653, 994.42471789, 3.14423556, -17.68296229, 0.79195877],
                          [2633.26416638, 292.67774166, 3785.20337309, 3.60715947, 2348.23810053, 274.94675522, -23.39908428, 0.77862948],
                          [3098.75567908, 28.47334134, 2895.85282881, 2.97318649, 3547.83278249, 2114.19931727, -68.57292675, 1.14634821]])
distance = [1.0, 1.1, 1.2, 1.3, 1.4, 1.5, 1.6, 1.7, 1.8, 1.9, 2.0, 2.2, 2.4, 2.6, 2.8, 3.0, 3.2, 3.4, 3.6, 3.8, 4.0, 4.2, 4.4, 4.6, 4.8, 5.0, 5.2, 5.4, 5.6, 5.8, 6.0, 6.2, 6.4, 6.6, 6.8, 7.0]
```

```python
x_data = np.array(distance[5:])
y_data = np.array(gege[5:])
result = minimize(objective_function, initial_guess[0], args=(x_data, y_data), method='BFGS')
fitted_params = result.x
print("0.75 1.5", end=" ")
for i, param in enumerate(fitted_params):
    print(np.round(param, 8), end=" ")

plt.scatter(x_data, y_data, label="Data")
x_func = np.linspace(1.5, 7, 100)
plt.plot(x_func, zbl_function(x_func, fitted_params), label="Fitted Curve")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.tight_layout()
plt.show()
```
